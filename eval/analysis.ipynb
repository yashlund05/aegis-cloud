# Aegis: Phase 9 Evaluation & Ablation Study Analysis
### Coupling Quantile Workload Forecasting with Energy-Aware Scheduling and Autoscaling

This notebook evaluates the 4 experimental configurations:
1. **`stock_hpa`**: Standard Kubernetes Horizontal Pod Autoscaler baseline.
2. **`forecast_only`**: Predictive quantile autoscaling without placement or node power management.
3. **`forecast_placement`**: Predictive autoscaling + energy-aware pod placement (CP-SAT/Go scheduler).
4. **`full_aegis`**: Complete closed loop: Forecasting + Joint Placement + Active Node Power Management.

```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt

# Load ablation benchmark results
report_path = 'ablation_results.json'
if not os.path.exists(report_path):
    report_path = os.path.join('eval', 'ablation_results.json')

with open(report_path, 'r') as f:
    data = json.load(f)

df_results = pd.DataFrame(data['configurations']).T
print('=== ABLATION BENCHMARK RESULTS ===')
df_results[['energy_kwh', 'slo_violations', 'slo_violation_rate_pct', 'scaling_churn', 'mean_active_nodes']]
```

```python
# Chart 1: Energy Consumption Comparison
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

df_results['energy_kwh'].plot(kind='bar', ax=ax[0], color=['#e74c3c', '#f39c12', '#3498db', '#2ecc71'])
ax[0].set_title('Total Cluster Energy Consumption (kWh)')
ax[0].set_ylabel('Energy (kWh)')
ax[0].set_xticklabels(df_results.index, rotation=25)

# Chart 2: SLO Violations
df_results['slo_violations'].plot(kind='bar', ax=ax[1], color=['#e74c3c', '#2ecc71', '#2ecc71', '#2ecc71'])
ax[1].set_title('SLO Violations Count (Breaches)')
ax[1].set_ylabel('Violations')
ax[1].set_xticklabels(df_results.index, rotation=25)

plt.tight_layout()
plt.show()
```
